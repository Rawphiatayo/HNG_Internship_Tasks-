```python
def clean_name(name):
    if pd.isna(name):
        return ""
    name = str(name).strip()
    # Remove @ at start
    name = re.sub(r'^@+', '', name)
    # Remove common emoji suffixes like 1️⃣, 2️⃣, 3️⃣, etc.
    name = re.sub(r'[0-9]️⃣', '', name)
    # Remove other emojis or special suffixes (optional)
    name = re.sub(r'\s*[0-9]+[️⃣]*\s*$', '', name)  # trailing numbers
    name = ' '.join(name.split())  # normalize spaces
    return name.lower()

new_df['Name_Clean'] = new_df['Slack Display Name'].apply(clean_name)
```

```python
def first_non_null(series):
    non_null = series.dropna()
    if non_null.empty:
        return None
    # Prefer non-"No Score", non-empty, non-placeholder values
    meaningful = non_null[~non_null.astype(str).isin(["", "No Score", "nan"])]
    if not meaningful.empty:
        return meaningful.iloc[0]
    return non_null.iloc[0]
```

```python
clean_df = new_df.groupby('Person_ID').agg(first_non_null)
```

```python
def score_grade(score):
    if pd.isnull(score):
        return 'No Score'
    elif score >= 8:
        return 'Distinction'
    elif score >= 7:
        return 'Second Class Upper'
    elif score >= 6:
        return 'Third Class'
    elif score >= 5:
        return 'Let my people go'
    else:
        return 'Fail'
```
