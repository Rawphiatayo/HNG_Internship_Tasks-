```python
import re

def clean_slack_name(name):
    if pd.isna(name): 
        return ""
    name = str(name).strip()
    
    # Remove @ prefix
    if name.startswith('@'):
        name = name[1:].strip()
    
    # Remove trailing numbers, emojis, symbols like 1️⃣, 2️⃣, etc.
    # Also remove " (1)", " [1]", etc.
    name = re.sub(r'[0-9]+️⃣|\s*\([0-9]+\)|\s*\[[0-9]+\]|\s*[1-9]$', '', name)
    name = re.sub(r'\s*;.*', '', name)  # e.g., "FaithA ;)"
    name = re.sub(r'\s+\d+$', '', name)  # e.g., "FaithCK 4"
    
    # Strip again
    name = name.strip()
    return name
new_df['Slack Display Name'] = new_df['Slack Display Name'].apply(clean_slack_name)
```
