```python
import pandas as pd

# Convert only numeric-like values to integers and label them
def format_year(val):
    try:
        # Try to convert numeric strings/floats like '2024.0' or 2024
        year_int = int(float(val))
        return f"Year {year_int}"
    except (ValueError, TypeError):
        # If it's not numeric, keep it unchanged
        return val

finalbank['Year'] = finalbank['Year'].apply(format_year)

finalbank.head()

```

```python
# Step 1: Load your existing CSV
df = pd.read_csv("keywords.csv")
keywords = df["Keyword Text"].tolist()
# Step 2: Define Nigerian banks to look for
banks = [
    "accessbank", "gtbank", "zenithbank", "firstbank", "uba",
    "polarisbank", "fcmb", "unionbank", "sterlingbank",
    "fidelitybank", "stanbicibtc", "providusbank", "keystonebank",
    "jaizbank", "tajbank", "wema", "globank"
]
api_key = "4a0ea474f3594ea972db084760235e10e30718a43a9bbccd79425dcfb4e7c736" 
results = []
```

```python
# Step 3: Scrape Google search results
api_key = "4a0ea474f3594ea972db084760235e10e30718a43a9bbccd79425dcfb4e7c736" 
def fetch_data(kw):
    params = {
        "engine": "google",
        "q": f"{kw} Nigeria",
        "api_key": api_key,
        "num": "10"}
    search = GoogleSearch(params)
    data = search.get_dict()
    output = []

    if "organic_results" in data:
        for idx, item in enumerate(data["organic_results"], start=1):
            link = item.get("link", "").lower()
            for bank in banks:
                if bank in link:
                    output.append({
                        "Keyword": kw,
                        "Bank": bank,
                        "Rank Position": idx
                    })
    return output
# Use ThreadPoolExecutor for parallel requests
with ThreadPoolExecutor(max_workers=5) as executor:
    future_to_kw = {executor.submit(fetch_data, kw): kw for kw in keywords}
    for future in as_completed(future_to_kw):
        try:
            results.extend(future.result())
        except Exception as e:
            print(f"Error with {future_to_kw[future]}: {e}")

# Save results
rank_df = pd.DataFrame(results)
rank_df.to_csv("bank_rank_results_fast.csv", index=False)
print("Fast bank ranking data saved successfully!")
```
