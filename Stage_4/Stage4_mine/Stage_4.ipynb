```python
df_national = pd.read_csv('data/raw/nigeria_national_production.csv')
df_states = pd.read_csv('data/raw/state_production_data.csv')
df_companies = pd.read_csv('data/raw/company_benchmarks.csv')
df_costs = pd.read_csv('data/raw/cost_structure.csv')
df_revenue = pd.read_csv('data/raw/revenue_assumptions.csv')
df_climate = pd.read_csv('data/raw/climate_data.csv')
```

```python
print("\n1. CHECKING FOR MISSING VALUES")
datasets = {
    'National Data': df_national,
    'State Data': df_states,
    'Company Benchmarks': df_companies,
    'Cost Structure': df_costs,
    'Revenue Assumptions': df_revenue,
    'Climate Data': df_climate
}

for name, df in datasets.items():
    missing = df.isnull().sum().sum()
    print(f"{name}: {missing} missing values")
```

```python
print("\n2. DATA TYPE CONVERSIONS")
numeric_cols_states = ['Area_Ha', 'Production_MT', 'Yield_MT_per_Ha', 
                       'Number_of_Mills', 'Average_Rainfall_mm', 
                       'Land_Cost_Naira_per_Ha', 'Infrastructure_Score']
for col in numeric_cols_states:
    df_states[col] = pd.to_numeric(df_states[col], errors='coerce')

print("✓ Converted state data numeric columns")
```

```python
print("\n3. CREATING MERGED DATASET")
# Merge state data with climate data
df_merged = df_states.merge(df_climate, on='State', how='left', suffixes=('', '_climate'))

# Handle duplicate rainfall columns
if 'Average_Rainfall_mm_climate' in df_merged.columns:
    df_merged['Average_Rainfall_mm'] = df_merged['Average_Rainfall_mm_climate']
    df_merged.drop('Average_Rainfall_mm_climate', axis=1, inplace=True)

print(f"✓ Merged dataset shape: {df_merged.shape}")
print(f"✓ Columns: {', '.join(df_merged.columns)}")
```

```python
print("\n4. CALCULATING COST SCENARIOS")

# Capital costs (amortized over 25 years for oil palm lifecycle)
capital_costs = df_costs[df_costs['Cost_Type'] == 'Capital']
operating_costs = df_costs[df_costs['Cost_Type'] == 'Operating']

cost_scenarios = {}
for scenario in ['Conservative', 'Base', 'Optimistic']:
    col_name = f'{scenario}_Naira_per_Ha'
    
    # Capital costs amortized
    capital_amortized = capital_costs[col_name].sum() / 25
    
    # Operating costs (annual)
    operating_annual = operating_costs[col_name].sum()
    
    # Total annual cost per hectare
    total_annual = capital_amortized + operating_annual
    
    cost_scenarios[scenario] = {
        'Capital_Amortized': capital_amortized,
        'Operating_Annual': operating_annual,
        'Total_Annual_per_Ha': total_annual
    }
    
    print(f"\n{scenario} Scenario:")
    print(f"  Capital (amortized): ₦{capital_amortized:,.0f}/ha/year")
    print(f"  Operating (annual): ₦{operating_annual:,.0f}/ha/year")
    print(f"  TOTAL: ₦{total_annual:,.0f}/ha/year")

# Save cost scenarios
df_cost_scenarios = pd.DataFrame(cost_scenarios).T
df_cost_scenarios.reset_index(inplace=True)
df_cost_scenarios.rename(columns={'index': 'Scenario'}, inplace=True)
df_cost_scenarios.to_csv('data/cleaned/cost_scenarios_annual.csv', index=False)
```

```python
print("\n5. CALCULATING REVENUE SCENARIOS")

for idx, row in df_revenue.iterrows():
    scenario = row['Scenario']
    
    # CPO revenue
    cpo_mt = row['FFB_Yield_MT_per_Ha'] * row['Oil_Extraction_Rate']
    cpo_revenue = cpo_mt * row['CPO_Price_Naira_per_MT']
    
    # Kernel revenue
    kernel_mt = row['FFB_Yield_MT_per_Ha'] * row['Kernel_Yield_Rate']
    kernel_revenue = kernel_mt * row['Kernel_Price_Naira_per_MT']
    
    # Total revenue
    total_revenue = cpo_revenue + kernel_revenue
    
    df_revenue.loc[idx, 'CPO_MT_per_Ha'] = cpo_mt
    df_revenue.loc[idx, 'CPO_Revenue_Naira'] = cpo_revenue
    df_revenue.loc[idx, 'Kernel_MT_per_Ha'] = kernel_mt
    df_revenue.loc[idx, 'Kernel_Revenue_Naira'] = kernel_revenue
    df_revenue.loc[idx, 'Total_Revenue_per_Ha'] = total_revenue
    
    print(f"\n{scenario}:")
    print(f"  CPO: {cpo_mt:.2f} MT/ha → ₦{cpo_revenue:,.0f}/ha")
    print(f"  Kernel: {kernel_mt:.2f} MT/ha → ₦{kernel_revenue:,.0f}/ha")
    print(f"  TOTAL REVENUE: ₦{total_revenue:,.0f}/ha")

# Save updated revenue data
df_revenue.to_csv('data/cleaned/revenue_scenarios_calculated.csv', index=False)
```

```python
print("PALM OIL PROFITABILITY ANALYSIS - NIGERIA")

# Load cleaned data
df_states = pd.read_csv('data/cleaned/state_merged_data.csv')
df_costs = pd.read_csv('data/cleaned/cost_scenarios_annual.csv')
df_revenue = pd.read_csv('data/cleaned/revenue_scenarios_calculated.csv')
df_companies = pd.read_csv('data/cleaned/company_benchmarks_clean.csv')
```

```python
# Calculate profit for each scenario
profitability_scenarios = []

for idx, rev_row in df_revenue.iterrows():
    scenario = rev_row['Scenario']
    cost_row = df_costs[df_costs['Scenario'] == scenario].iloc[0]
    
    revenue_per_ha = rev_row['Total_Revenue_per_Ha']
    cost_per_ha = cost_row['Total_Annual_per_Ha']
    profit_per_ha = revenue_per_ha - cost_per_ha
    profit_margin = (profit_per_ha / revenue_per_ha) * 100 if revenue_per_ha > 0 else 0
    
    profitability_scenarios.append({
        'Scenario': scenario,
        'Revenue_per_Ha': revenue_per_ha,
        'Cost_per_Ha': cost_per_ha,
        'Profit_per_Ha': profit_per_ha,
        'Profit_Margin_Percent': profit_margin
    })
    
    print(f"\n{scenario.upper()} SCENARIO:")
    print(f"  Revenue per ha: ₦{revenue_per_ha:,.0f}")
    print(f"  Cost per ha:    ₦{cost_per_ha:,.0f}")
    print(f"  Profit per ha:  ₦{profit_per_ha:,.0f}")
    print(f"  Profit margin:  {profit_margin:.1f}%")

df_profitability = pd.DataFrame(profitability_scenarios)
df_profitability.to_csv('data/cleaned/profitability_by_scenario.csv', index=False)

```

```python
# Break-even analysis for ₦1 billion profit
target_profit = 1_000_000_000  # ₦1 billion

breakeven_analysis = []

for idx, row in df_profitability.iterrows():
    scenario = row['Scenario']
    profit_per_ha = row['Profit_per_Ha']
    
    if profit_per_ha > 0:
        hectares_needed = target_profit / profit_per_ha
        breakeven_analysis.append({
            'Scenario': scenario,
            'Profit_per_Ha': profit_per_ha,
            'Hectares_for_1B_Profit': hectares_needed,
            'Acres_for_1B_Profit': hectares_needed * 2.471  # Convert to acres
        })
        
        print(f"\n{scenario.upper()}:")
        print(f"  Profit per hectare: ₦{profit_per_ha:,.0f}")
        print(f"  Hectares needed for ₦1B profit: {hectares_needed:,.0f} ha")
        print(f"  Acres needed: {hectares_needed * 2.471:,.0f} acres")
    else:
        print(f"\n{scenario.upper()}: NOT PROFITABLE (Loss: ₦{abs(profit_per_ha):,.0f}/ha)")
df_breakeven = pd.DataFrame(breakeven_analysis)
df_breakeven.to_csv('analysis/breakeven_analysis.csv', index=False)
```

```python
# Calculate total investment needed for each scenario
investment_analysis = []

for idx, row in df_breakeven.iterrows():
    scenario = row['Scenario']
    hectares = row['Hectares_for_1B_Profit']
    
    # Get cost data for this scenario
    cost_scenario = df_costs[df_costs['Scenario'] == scenario].iloc[0]
    
    # Capital costs (not amortized - full upfront)
    df_cost_structure = pd.read_csv('data/cleaned/cost_structure_clean.csv')
    capital_costs = df_cost_structure[df_cost_structure['Cost_Type'] == 'Capital']
    
    col_name = f'{scenario}_Naira_per_Ha'
    total_capital_per_ha = capital_costs[col_name].sum()
    
    # Total investment
    total_investment = total_capital_per_ha * hectares
    
    # First year operating costs
    operating_per_ha = cost_scenario['Operating_Annual']
    first_year_operating = operating_per_ha * hectares
    
    # Total upfront investment
    total_upfront = total_investment + first_year_operating
    
    investment_analysis.append({
        'Scenario': scenario,
        'Hectares': hectares,
        'Capital_per_Ha': total_capital_per_ha,
        'Total_Capital_Investment': total_investment,
        'First_Year_Operating_Costs': first_year_operating,
        'Total_Upfront_Investment': total_upfront
    })
    
    print(f"\n{scenario.upper()}:")
    print(f"  Plantation size: {hectares:,.0f} hectares")
    print(f"  Capital cost per ha: ₦{total_capital_per_ha:,.0f}")
    print(f"  Total capital investment: ₦{total_investment:,.0f}")
    print(f"  First year operating: ₦{first_year_operating:,.0f}")
    print(f"  TOTAL UPFRONT: ₦{total_upfront:,.0f}")

df_investment = pd.DataFrame(investment_analysis)
df_investment.to_csv('analysis/investment_analysis.csv', index=False)
```

```python
# ROI and Payback Period Analysis

roi_analysis = []

for idx, inv_row in df_investment.iterrows():
    scenario = inv_row['Scenario']
    total_investment = inv_row['Total_Upfront_Investment']
    
    # Annual profit (once mature - year 4+)
    profit_row = df_profitability[df_profitability['Scenario'] == scenario].iloc[0]
    hectares = inv_row['Hectares']
    annual_profit_mature = profit_row['Profit_per_Ha'] * hectares
    
    # ROI calculation
    roi_percent = (annual_profit_mature / total_investment) * 100
    
    # Payback period (simplified - not accounting for immature years)
    payback_years = total_investment / annual_profit_mature if annual_profit_mature > 0 else 0
    
    # Adjusted payback (accounting for 3 years of low/no production)
    # Years 1-3: minimal production, Year 4+: full production
    payback_adjusted = payback_years + 3
    
    roi_analysis.append({
        'Scenario': scenario,
        'Total_Investment': total_investment,
        'Annual_Profit_Mature': annual_profit_mature,
        'ROI_Percent': roi_percent,
        'Payback_Years_Simple': payback_years,
        'Payback_Years_Adjusted': payback_adjusted
    })
    
    print(f"\n{scenario.upper()}:")
    print(f"  Total investment: ₦{total_investment:,.0f}")
    print(f"  Annual profit (mature): ₦{annual_profit_mature:,.0f}")
    print(f"  ROI: {roi_percent:.1f}%")
    print(f"  Payback period: {payback_adjusted:.1f} years (incl. maturation)")

df_roi = pd.DataFrame(roi_analysis)
df_roi.to_csv('analysis/roi_payback_analysis.csv', index=False)
```

```python
# Use BASE scenario for state comparison
base_revenue = df_revenue[df_revenue['Scenario'] == 'Base'].iloc[0]
base_cost = df_costs[df_costs['Scenario'] == 'Base'].iloc[0]

state_profitability = []

for idx, state_row in df_states.iterrows():
    state = state_row['State']
    
    # Adjust yield based on rainfall (optimal: 2000-3000mm)
    rainfall = state_row['Average_Rainfall_mm']
    if rainfall < 2000:
        yield_factor = 0.85
    elif rainfall > 3000:
        yield_factor = 0.90
    else:
        yield_factor = 1.0
    
    # Adjust for infrastructure
    infra_score = state_row['Infrastructure_Score']
    infra_factor = 0.85 + (infra_score / 10) * 0.15  # 0.85 to 1.0
    
    # Combined adjustment
    state_factor = yield_factor * infra_factor
    
    # Calculate state-specific metrics
    revenue_per_ha = base_revenue['Total_Revenue_per_Ha'] * state_factor
    cost_per_ha = base_cost['Total_Annual_per_Ha']
    
    # Adjust costs based on land cost differential
    avg_land_cost = df_states['Land_Cost_Naira_per_Ha'].mean()
    land_cost_factor = state_row['Land_Cost_Naira_per_Ha'] / avg_land_cost
    cost_per_ha *= land_cost_factor
    
    profit_per_ha = revenue_per_ha - cost_per_ha
    hectares_for_1b = target_profit / profit_per_ha if profit_per_ha > 0 else 0
    
    state_profitability.append({
        'State': state,
        'Yield_Factor': state_factor,
        'Revenue_per_Ha': revenue_per_ha,
        'Cost_per_Ha': cost_per_ha,
        'Profit_per_Ha': profit_per_ha,
        'Hectares_for_1B_Profit': hectares_for_1b,
        'Rainfall_mm': rainfall,
        'Infrastructure_Score': infra_score,
        'Land_Cost_per_Ha': state_row['Land_Cost_Naira_per_Ha']
    })
    
    print(f"\n{state.upper()}:")
    print(f"  Yield factor: {state_factor:.2f}")
    print(f"  Profit per ha: ₦{profit_per_ha:,.0f}")
    print(f"  Hectares for ₦1B: {hectares_for_1b:,.0f} ha")

df_state_profit = pd.DataFrame(state_profitability)
df_state_profit = df_state_profit.sort_values('Profit_per_Ha', ascending=False)
df_state_profit.to_csv('analysis/state_profitability_ranking.csv', index=False)
```

```python
# Top 3 state recommendations
top_3_states = df_state_profit.head(3)

recommendations = []
for idx, state in top_3_states.iterrows():
    rank = idx + 1
    
    # Investment calculation for this state
    hectares = state['Hectares_for_1B_Profit']
    
    # Use base scenario capital costs
    df_cost_structure = pd.read_csv('data/cleaned/cost_structure_clean.csv')
    capital_costs = df_cost_structure[df_cost_structure['Cost_Type'] == 'Capital']
    capital_per_ha = capital_costs['Base_Naira_per_Ha'].sum()
    
    total_investment = (capital_per_ha * hectares) + (state['Cost_per_Ha'] * hectares)
    
    recommendations.append({
        'Rank': rank,
        'State': state['State'],
        'Profit_per_Ha': state['Profit_per_Ha'],
        'Hectares_Needed': hectares,
        'Total_Investment': total_investment,
        'Key_Advantage': 'High rainfall & infrastructure' if state['Rainfall_mm'] > 2500 
                         else 'Established industry & access'
    })
    
    print(f"\nRANK #{rank}: {state['State'].upper()}")
    print(f"  Profit per hectare: ₦{state['Profit_per_Ha']:,.0f}")
    print(f"  Hectares needed: {hectares:,.0f} ha")
    print(f"  Investment required: ₦{total_investment:,.0f}")
    print(f"  Infrastructure score: {state['Infrastructure_Score']}/10")

df_recommendations = pd.DataFrame(recommendations)
df_recommendations.to_csv('analysis/top_3_recommendations.csv', index=False)
```
